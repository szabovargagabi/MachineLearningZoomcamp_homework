```python
X=car_Asia[['vehicle_weight','model_year']].values
```

```python
XTX=(X.T).dot(X)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
XTX_inv = inv(XTX)
```

```python
w=(XTX_inv.dot(X.T)).dot(y)
```
